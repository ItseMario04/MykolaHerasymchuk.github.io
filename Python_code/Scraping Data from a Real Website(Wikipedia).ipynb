# Scraping Data from a Real Website + Pandas

```python
from bs4 import BeautifulSoup
import requests
```

```python
url = "https://en.wikipedia.org/wiki/List_of_largest_companies_by_revenue"

headers = {
    'User-Agent': 'LearningScraperApp/1.0 (cracer2004@gmail.com)'
}

page = requests.get(url, headers=headers)

if page.status_code == 200:
    soup = BeautifulSoup(page.text, 'html.parser')
    print(soup.title.text) 
else:
    print(f"Помилка доступу: {page.status_code}")
```

```python
soup = BeautifulSoup(page.text, 'html')
print(soup.prettify())
```

```python
soup.find('table', class_ = "wikitable sortable sticky-header-multi sort-under")
```

```python
table = soup.find('table', class_ = "wikitable sortable sticky-header-multi sort-under")
```

```python
world_titles = table.find_all('th')
```

```python
world_titles
```

```python
world_table_title = [title.text.strip() for title in world_titles]
print(world_table_title)
```

```python
world_table_titles = (world_table_title[1:7])
world_table_titles[5] = 'Headquarters'
world_table_ranks = (world_table_title[10:])
print(world_table_titles)
```

```python
import pandas as pd
```

```python
data_frame = pd.DataFrame(columns = world_table_titles)

data_frame
```

```python
rank_rows = pd.DataFrame(columns = world_table_ranks)
rank_rows
```

```python
column_data = table.find_all('tr')
column_data
```

```python
for row in column_data[2:]:
    row_data = row.find_all('td')    
    individual_row_dat = [data.text.strip() for data in row_data]
    individual_row_data = individual_row_dat[:6]
    #print(individual_row_data)
    length = len(data_frame)
    data_frame.loc[length] = individual_row_data
```
